## Task 5

```python
ax = df.plot(
    kind='scatter',
    x='age',
    y='charges',
    figsize=(10, 5)
)

plt.title(
    "Patient Age vs. Insurance Charges",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Age (years)")
plt.ylabel("Annual Charges (USD)")
plt.show()
```

The font is an appropriate size.  
It feels right that the title is larger to grab the reader's attention to the topic immediately.


## Task 6

```python
ax = df.plot(
    kind='scatter',
    x='age',
    y='charges',
    figsize=(10, 5)
)

plt.title(
    "Patient Age vs. Insurance Charges",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Age (years)")
plt.ylabel("Annual Charges (USD)")
plt.ylim(0,70000)
plt.show()
```

(5000,30000): This setting cut off a good portion of the data.  
This would have been avoided by starting at 0. Most charts start at zero, because it is easier to understand the relationship. If the data is clustered at a certain value range, you would not need to start at zero and it paints a simple picture.

## Task 7

```python
region_avg = df.groupby('region')['charges'].mean()
ax = region_avg.plot(
    kind='bar', 
    figsize=(8, 5),
    title="Mean Charges by Region",
    xlabel="Region",
    ylabel="Charges (USD)",
    color="LightPink")
plt.xticks(rotation=45, ha='right')
plt.show()
```
