(5000,30000): This setting cut off a good portion of the data.  
This would have been avoided by starting at 0. Most charts start at zero, because it is easier to understand the relationship. If the data is clustered at a certain value range, you would not need to start at zero and it paints a simple picture.

## Task 7

```python
ax = df.plot(
    kind='scatter',
    x='bmi',
    y='charges',
    figsize=(10, 5),
    alpha=0.4
)

plt.title("BMI vs Charges")
plt.xlabel("BMI")
plt.ylabel("Charges (USD)")
plt.show()
```

alpha = 0.4 shows the data clearly and allows you to view the overlapping data.  
A very dense region indicates many overlapping data points. A solid dot scatter plot would make those look like one data point.

## Task 12

```python
region_avg = df.groupby('region')['charges'].mean()
ax = region_avg.plot(
    kind='bar', 
    figsize=(8, 5),
    title="Mean Charges by Region",
    xlabel="Region",
    ylabel="Charges (USD)",
    color="steelblue",
    edgecolor="#501CA3")
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', linewidth=2.0)

plt.show()
```

The horizontal line is average charges

## Task 20

```python
region_avg = df.groupby('region')['charges'].mean()
```

```python
max_region = region_avg.idxmin()
colors = [
    'red' if r == max_region else 'lightgray'
    for r in region_avg.index
]
```

```python
ax = region_avg.plot(
    kind='bar',
    figsize=(8, 5),
    color=colors,
    title="Average Regional Cost",
    xlabel="Region",
    ylabel="Charges (USD)"
)
```

The leader should notice the highlighted color first. Very effective!

## Task 21

```python
plt.style.use('classic')
plt.figure(figsize=(11, 6))

ax = sns.scatterplot(
    data=df,
    x='bmi',
    y='charges',
    hue="smoker",
    palette="colorblind",
    alpha=0.45
)

plt.title("BMI vs Charges", weight="bold", fontsize=15)
plt.xlabel("BMI", fontsize=12)
plt.ylabel("Charges (USD)", fontsize=12)
plt.axvline(
    x=df['bmi'].mean(),
    color='red',
    linestyle='--',
    linewidth=1.5
)
plt.axhline(
    y=df['charges'].mean(),
    color='red',
    linestyle='--',
    linewidth=1.5
)
plt.text((df['bmi'].mean() + 0.5), 60000, "Mean BMI")
plt.text((47), df['charges'].mean() + 950, "Mean Charges")
plt.grid(axis='y', linestyle='--', linewidth=0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.legend(loc="upper left", title="Smoker")
plt.savefig("bmi_vs_charges_polished.png", dpi=150)
plt.show()
```
